```python
# import libraries
import joblib
import pandas as pd
```

```python
# load trained model
model = joblib.load('../model/bangalore_house_price_model.pkl')
```

```python
# test data
test_data = pd.DataFrame([{
    'area_type': 'Super built-up  Area',
    'availability': 'Ready To Move',
    'location': 'whitefield',
    'size': 2,                 # bhk
    'total_sqft': 1200.0,
    'bath': 2.0,
    'balcony': 1.0
}])
```

```python
# make prediction
predicted_price = model.predict(test_data)
print(f"Predicted House Price (in Lakhs): {predicted_price[0]:.2f}")
```
